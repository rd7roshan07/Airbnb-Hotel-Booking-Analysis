```python
from os import dup
#data quality check
duplicate_count = df.duplicated().sum()
print('Duplicate Rows',duplicate_count)
```

```python
df = df.drop_duplicates().copy()
print('Duplicates remaining',df.duplicated().sum())
```

```python
missing = pd.DataFrame({
    'Missing Count': df.isna().sum(),
    'Missing %': (df.isna().mean() * 100).round(2)
})
missing = missing[missing['Missing Count'] > 0]
missing.sort_values('Missing Count', ascending=False)
```

```python
# Categorical columns: missing means we do not know the category.
categorical_columns = [
    'NAME', 'host_identity_verified', 'host name',
    'neighbourhood group', 'neighbourhood', 'country',
    'country code', 'cancellation_policy', 'house_rules'
]
for col in categorical_columns:
    df[col] = df[col].fillna('Unknown')
```

```python
# Review counts: if there is no recorded review count, treat it as zero.
df['number of reviews'] = df['number of reviews'].fillna(0)
df['reviews per month'] = df['reviews per month'].fillna(0)
```

```python
# price is important for analysis , so remove rows where price is missing.
df =df.dropna(subset=['price']).copy()
```

```python
df['neighbourhood group'] = df['neighbourhood group'].replace({
    'manhatan': 'Manhattan',
    'brookln': 'Brooklyn'
})
df['neighbourhood group'].value_counts()
```

```python
df.loc[~df['minimum nights'].between(1, 365), 'minimum nights'] = np.nan
df.loc[~df['availability 365'].between(0, 365), 'availability 365'] = np.nan
df['minimum nights'] = df['minimum nights'].fillna(df['minimum nights'].median())
df['availability 365'] = df['availability 365'].fillna(df['availability 365'].median())
```

```python
# Numerical columns where median imputation is reasonable for this exercise.
df['Construction year'] = df['Construction year'].fillna(df['Construction year'].median())
df['service fee'] = df['service fee'].fillna(df['service fee'].median())
df['calculated host listings count'] = df['calculated host listings count'].fillna(
    df['calculated host listings count'].median()
)
```

```python
missing_after = pd.DataFrame({
    'Missing Count': df.isna().sum(),
    'Missing %': (df.isna().mean() * 100).round(2)
})
missing_after = missing_after[missing_after['Missing Count'] > 0]
missing_after.sort_values('Missing Count', ascending=False)
```

```python
#Analyze
borough_counts = df['neighbourhood group'].value_counts()
borough_counts
```

```python
room_counts = df['room type' ].value_counts()
room_counts
```

```python
average_price = df['price'].mean()
median_price = df['price'].median()
print(f'Average price: ${average_price:,.2f}')
print(f'Median price: ${median_price:,.2f}')
```

```python
# Chart 1: Listings by borough
plt.figure(figsize=(8, 5))
borough_counts.plot(kind='bar')
plt.title('Airbnb Listings by Borough')
plt.xlabel('Borough')
plt.ylabel('Number of Listings')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Chart 2: Listings by room type
plt.figure(figsize=(8, 5))
room_counts.plot(kind='bar')
plt.title('Airbnb Listings by Room Type')
plt.xlabel('Room Type')
plt.ylabel('Number of Listings')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Chart 3: Average price by room type
avg_price_room = df.groupby('room type')['price'].mean()
plt.figure(figsize=(8, 5))
avg_price_room.plot(kind='bar')
plt.title('Average Price by Room Type')
plt.xlabel('Room Type')
plt.ylabel('Average Price ($)')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='number of reviews', y='price', data=df, alpha=0.6)
plt.title('Correlation between Price and Number of Reviews')
plt.xlabel('Number of Reviews')
plt.ylabel('Price')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 7))
sns.boxplot(x='neighbourhood group', y='price', data=df)
plt.title('Price Distribution by Neighbourhood Group')
plt.xlabel('Neighbourhood Group')
plt.ylabel('Price')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
avg_price_borough = df.groupby('neighbourhood group')['price'].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=avg_price_borough.index, y=avg_price_borough.values, hue=avg_price_borough.index, palette='viridis', legend=False)
plt.title('Average Price by Neighbourhood Group')
plt.xlabel('Neighbourhood Group')
plt.ylabel('Average Price ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Now let's visualize the relationship between 'price' and 'minimum nights'.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='minimum nights', y='price', data=df, alpha=0.6)
plt.title('Relationship between Price and Minimum Nights')
plt.xlabel('Minimum Nights')
plt.ylabel('Price')
plt.tight_layout()
plt.show()
```

Finally, let's look at the top 10 hosts based on the number of listings.

```python
top_hosts = df['host name'].value_counts().head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x=top_hosts.index, y=top_hosts.values, hue=top_hosts.index, palette='magma', legend=False)
plt.title('Top 10 Hosts by Number of Listings')
plt.xlabel('Host Name')
plt.ylabel('Number of Listings')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
